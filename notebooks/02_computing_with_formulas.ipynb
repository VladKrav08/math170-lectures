### Every variable has a type

Every value in Python has a **type**, and the type controls what you can do with the
value. For now you only need four:

- **`int`** — a whole number, no decimal point: `7`, `-3`, `0`.
- **`float`** — a number *with* a decimal point: `5.0`, `3.14`, `-0.5`. Note that
  `5` is an `int` but `5.0` is a `float`, even though they look "equal".
- **`str`** — a *string*, i.e. text, always written inside quotes: `"Hello"`, `'42'`.
  Careful: `"42"` is text, not the number 42.
- **`bool`** — a *boolean*, either `True` or `False` (the result of a comparison
  like `a == 5`).

You never declare the type yourself — Python figures it out from how you write the
value. You can always ask what type a value has with the built-in `type()` function,
e.g. `type(7)` gives `<class 'int'>`.

**Integer division.** Predict both values, then run.

```python
print(5/100)
print(5.0/100)
```

In Python 3 both give `0.05`. In Python 2 (and some other languages) `5/100` would be `0`: between two integers, `/` performs *integer division*, so the result is the whole-number part of the quotient — 5 divided by 100 is 0.05, which becomes `0`. In a comment, note the habit this suggests: write quantities used in division as floats (e.g. `r = 5.0`) so your code behaves the same if you port it to another language.
