### The `while` loop

Python's most general loop is the `while` loop:

```python
while condition:
    <statement 1>
    <statement 2>
```

- The **condition** is a Python expression that is either true or false — a
  *Boolean expression* (more on those in 3.2). It is checked before every pass.
- The **body** is everything indented under the header, and it repeats as long as
  the condition holds.
- The loop ends at the first line that is no longer indented. Indentation is how
  Python groups statements into a block, so every line to be repeated must be
  indented, and all of them by exactly the same amount.

### Compound expressions

`and`, `or` and `not` combine Boolean expressions:

- `C1 and C2` is `True` if **both** conditions are `True`.
- `C1 or C2` is `True` if **at least one** is `True`.
- `not C` is `True` when `C` is `False`, and vice versa.

Go through the following examples by hand and predict the outcome, and then try
to run the code to obtain the result.

````python
x = 0;  y = 1.2
print(x >= 0 and y < 1)
print(x >= 0 or y < 1)
print(x > 0 or y > 1)
print(x > 0 or not y > 1)
print(-1 < x <= 0)            # same as -1 < x and x <= 0
print(not (x > 0 or y > 0))
````

# 3.4 Iterating Over a List with a `for` Loop

A `for` loop runs through the elements of a list and performs operations on each
one:

```python
for element in somelist:
    <statement 1>
    <statement 2>
```

On each pass the current element is stored in the variable `element` and the
indented block is executed. When the block finishes, the loop moves on to the
next element, and continues until there are no elements left — no stop condition
and no manual variable update needed.

Use a `for` loop when the list, i.e. the number of repetitions, is known in
advance; use a `while` loop when it is not.

### From `while` to `for`

Here is the investment example written as a `while` loop.

````python
# while-loop version
P = 100.0
r = 5.0
n = 0
while n <= 10:
    A = P * (1+r/100)**n
    print(n, A)
    n += 1
````
