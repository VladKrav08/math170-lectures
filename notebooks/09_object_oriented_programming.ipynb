# 9.1 Class Hierarchies and Inheritance

So far every class we wrote stood alone. **Object-oriented programming** is
programming with **class hierarchies**: families of closely related classes. A
**subclass** (child class) **inherits** the attributes and methods of its **base
class** (parent class, superclass), so code that the classes share is written
once, in the base class, and each subclass adds or changes only what is
different.

## Live coding

### A class for straight lines

Run the next cell. `Line` evaluates the straight line $y = c_0 + c_1x$, and its
method `table(L, R, n)` returns a table of `n` values for $L \le x \le R$.

````python
import numpy as np

class Line:
    def __init__(self, c0, c1):
        self.c0, self.c1 = c0, c1

    def __call__(self, x):
        return self.c0 + self.c1*x

    def table(self, L, R, n):
        """Return a table with n points for L <= x <= R."""
        s = ''
        for x in np.linspace(L, R, n):
            y = self(x)
            s += f'{x:12g} {y:12g}\n'
        return s
````

# 9.2 Example: Classes for Numerical Differentiation

A class hierarchy pays off when several classes differ in a single method. Here
each class is a different formula for the derivative of a function, and each

### A class hierarchy for numerical differentiation

**Objective.** We want to compute the derivative of a function `f` numerically,
using several different finite-difference formulas, each with a different
order of accuracy:

$$f'(x) = \frac{f(x+h) - f(x)}{h} + O(h) \quad \text{(forward)}$$
$$f'(x) = \frac{f(x) - f(x-h)}{h} + O(h) \quad \text{(backward)}$$
$$f'(x) = \frac{f(x+h) - f(x-h)}{2h} + O(h^2) \quad \text{(central, 2nd order)}$$
$$f'(x) = \frac{4}{3}\frac{f(x+h) - f(x-h)}{2h} - \frac{1}{3}\frac{f(x+2h) - f(x-2h)}{4h} + O(h^4) \quad \text{(central, 4th order)}$$
$$f'(x) = \frac{3}{2}\frac{f(x+h) - f(x-h)}{2h} - \frac{3}{5}\frac{f(x+2h) - f(x-2h)}{4h} + \frac{1}{10}\frac{f(x+3h) - f(x-3h)}{6h} + O(h^6) \quad \text{(central, 6th order)}$$
$$f'(x) = \frac{1}{h}\left(-\frac{1}{6}f(x+2h) + f(x+h) - \frac{1}{2}f(x) - \frac{1}{3}f(x-h)\right) + O(h^3) \quad \text{(mixed, 3rd order)}$$

# 9.3 Example: Classes for Numerical Integration

The same idea as in Section 9.2, now for integrals: what every method shares goes

### A class hierarchy for numerical integration

**Introduction.** Numerical integration methods approximate a definite integral
by a weighted sum of function values at a set of points:

$$\int_a^b f(x)\,dx \approx \sum_{i=0}^{n-1} w_i f(x_i).$$

Different methods are just different choices of the points $x_i$ and weights
$w_i$. As presented in the textbook:

- **Trapezoidal rule:** $x_i = a + ih$, with $h = (b-a)/(n-1)$, and
  $w_0 = w_{n-1} = h/2$, $w_i = h$ for $i \ne 0, n-1$.
- **Midpoint rule:** $x_i = a + \frac{h}{2} + ih$, with $h = (b-a)/n$, and
  $w_i = h$ for all $i$.
- **Simpson's rule:** $x_i = a + ih$, with $h = (b-a)/(n-1)$, and
  $w_0 = w_{n-1} = h/6$, $w_i = h/3$ for even $i$, $w_i = 2h/3$ for odd $i$.

**The superclass.** Since $\sum_i w_i f(x_i)$ is identical for every method, and
only the points/weights differ, the common part is placed in a superclass, and
each method's points/weights are computed in a subclass method
`construct_method`. You're given the superclass as-is:

````python
import numpy as np

class Integrator:
    def __init__(self, a, b, n):
        self.a, self.b, self.n = a, b, n
        self.points, self.weights = self.construct_method()

    def construct_method(self):
        raise NotImplementedError('no rule in class %s' % \
                                  self.__class__.__name__)

    def integrate(self, f):
        s = 0
        for i in range(len(self.weights)):
            s += self.weights[i]*f(self.points[i])
        return s
````
