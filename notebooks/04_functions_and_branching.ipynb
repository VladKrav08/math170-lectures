# 4.2 Function Arguments and Local Variables

Inside a function, the arguments are local variables like any other. Variables
defined in the main program are **global**: they can be used everywhere.
Variables defined inside a function are **local**: they exist only inside it.

````python
P = 100
r = 5.0
def amount(n):
    r = 4.0
    return P*(1+r/100)**n
print(amount(n=6))
print(r)

# YOUR CODE HERE -- note what you changed
````

After the function has ended, the local variable no longer exists (in
programming terms, it goes out of scope), whereas the global `r` is still there
and has its original value. If we actually want to change a global variable
inside a function, we must explicitly state so by using the keyword `global`.

*Aside: using `global` inside a function works, but it is not recommended
practice — just be aware it exists rather than reach for it. The better habit is
to keep a function self-contained: pass in whatever it needs as arguments or
define it as a local variable, and if the function needs to change a "global"
value, have it `return` the new value instead of reaching outside itself with
`global`.*

# 4.4 If-Tests for Branching the Program Flow

Many mathematical functions are defined piecewise, for example

$$f(x) = \begin{cases} \sin x, & 0 \le x \le \pi \\ 0, & \text{otherwise.} \end{cases}$$

To implement it, the program must *choose* what to do depending on `x`. This is
called **branching**, and the basic tool is the `if`-`else` block:

```python
if condition:
    <block of statements, executed if condition is True>
else:
    <block of statements, executed if condition is False>
<next statement, always executed>
```

The `else` part is optional.

# 4.5 Functions as Arguments to Functions

An argument to a Python function can be *any* Python object — including another
function. That is how we write general tools (derivatives, integrals, equation
solvers) that work for any $f(x)$.

The second derivative can be approximated by

$$f''(x) \approx \frac{f(x-h) - 2f(x) + f(x+h)}{h^2}.$$

Run this cell first.

````python
def diff2(f, x, h=1E-6):
    r = (f(x-h) - 2*f(x) + f(x+h))/float(h*h)
    return r
````

# 4.7 Writing Test Functions to Verify our Programs

A **test function** checks that one small unit of a program — usually one
function — does what it should. Writing tests takes time up front, but it saves
time overall: bugs are caught early, and a larger program can be built from
pieces you already know work.

### Known-answer tests

by hand. Run this cell first.

````python
def amount(n, P=100, r=5.0):
    return P*(1+r/100)**n
````

Python's `assert` statement checks a condition:

```python
assert condition, message
```

If `condition` is `True`, nothing happens. If it is `False`, the program stops
with an `AssertionError` and shows `message` (the message is optional). A test
function

- contains at least one `assert`, and may contain several;
- takes **no arguments** — whatever the tested function needs is defined as local
  variables inside the test;
- is named `test_<name of the function being tested>`.

### Comparing floats

With a 10 % rate, 100 grows to $100 \cdot 1.1 \cdot 1.1 = 121$ in two years.
Predict whether this test passes.

````python
# predict: ...
def test_amount_two_years():
    n, P, r = 2, 100, 10.0
    expected = 121.0
    computed = amount(n, P, r)
    success = computed == expected
    assert success, f'expected {expected}, got {computed}'

test_amount_two_years()
````

### Running tests automatically with `pytest`

In practice, tests live in `.py` files and are run by a tool called `pytest`.
From a terminal, `pytest myfile.py` finds and runs every function whose name
starts with `test_` — whether or not your code calls it — and prints a summary.
`pytest somefolder` also searches the subfolders for files named `test_*.py` or
`*_test.py`.
